# **UMAP Feature Space Visualization of Raw Data**

```python
import pandas as pd
df = pd.read_csv('/content/drive/MyDrive/Dataset/Final FV/COMBINE FILE-FV/DATASET 1.csv')

```

```python
from sklearn.utils import shuffle

df= shuffle(df, random_state=42)
```

```python
y=df['CLASS-LABEL']
```

```python
x=df.drop(['CLASS-LABEL'],axis=1)
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn import preprocessing
le = preprocessing.LabelEncoder()
x = x.apply(le.fit_transform)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
scaler.get_params()
x=scaler.fit_transform(x)
```

TE, NON TE Feature Space Visualization

```python
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.utils import shuffle
import umap
import plotly.express as px

# ================================
# Class label names (binary)
# ================================
class_names = ['TE', 'NON-TE']

# ================================
# Shuffle and scale data
# ================================
x, y = shuffle(x, y, random_state=42)
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)

# ================================
# PCA for noise reduction before UMAP
# ================================
print("Running PCA (50 components)...")
pca = PCA(n_components=50, random_state=42)
x_pca = pca.fit_transform(x_scaled)

# ================================
# UMAP in 2D
# ================================
print("Running 2D UMAP...")
umap_2d = umap.UMAP(
    n_neighbors=40,      # balance between local/global structure
    min_dist=0.25,       # tighter clusters
    n_components=2,
    metric='cosine',
    random_state=42
)
x_umap_2d = umap_2d.fit_transform(x_pca)

# ================================
# UMAP in 3D
# ================================
print("Running 3D UMAP...")
umap_3d = umap.UMAP(
    n_neighbors=40,
    min_dist=0.25,
    n_components=3,
    metric='cosine',
    random_state=42
)
x_umap_3d = umap_3d.fit_transform(x_pca)

# ================================
# Interactive 2D Plot (Plotly)
# ================================
df_2d = pd.DataFrame({
    "UMAP1": x_umap_2d[:, 0],
    "UMAP2": x_umap_2d[:, 1],
    "Label": [class_names[i] for i in y]
})

fig2d = px.scatter(
    df_2d, x="UMAP1", y="UMAP2",
    color="Label", title="2D Feature Space (UMAP)",
    color_discrete_sequence=["#1f77b4", "#d62728"],
    opacity=0.6, width=900, height=700
)

fig2d.update_traces(marker=dict(size=5, line=dict(width=0)))
fig2d.update_layout(
    font=dict(size=16),
    legend=dict(title="Class", orientation="v", yanchor="top", y=1, xanchor="left", x=1.02)
)
fig2d.show()

# ================================
# Interactive 3D Plot (Plotly)
# ================================
df_3d = pd.DataFrame({
    "UMAP1": x_umap_3d[:, 0],
    "UMAP2": x_umap_3d[:, 1],
    "UMAP3": x_umap_3d[:, 2],
    "Label": [class_names[i] for i in y]
})

fig3d = px.scatter_3d(
    df_3d, x="UMAP1", y="UMAP2", z="UMAP3",
    color="Label", title="3D Feature Space (UMAP)",
    opacity=0.6, width=900, height=700,
    color_discrete_sequence=["#1f77b4", "#d62728"],
)
fig3d.update_traces(marker=dict(size=3, line=dict(width=0)))
fig3d.update_layout(
    font=dict(size=16),
    legend=dict(title="Class", orientation="v", yanchor="top", y=1, xanchor="left", x=1.02)
)
fig3d.show()

# ================================
# Save plots
# ================================
fig2d.write_html("umap_2d_TE_vs_NONTE.html")
fig3d.write_html("umap_3d_TE_vs_NONTE.html")

print("✅ Saved interactive UMAP plots: umap_2d_TE_vs_NONTE.html and umap_3d_TE_vs_NONTE.html")
```

Orders Feature Space Visualization

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.utils import shuffle
from umap import UMAP
import plotly.express as px

# ================================
# Class label names
# ================================
class_names = ['LTR','LINE','SINE','CLASS II']

# ================================
# Shuffle data
# ================================
x, y = shuffle(x, y, random_state=42)

# ================================
# PCA (noise reduction before UMAP)
# ================================
print("Running PCA...")
pca = PCA(n_components=50, random_state=42)
x_pca = pca.fit_transform(x)

# ================================
# UMAP in 2D
# ================================
print("Running 2D UMAP...")
umap2d = UMAP(n_neighbors=30, min_dist=0.5, n_components=2, random_state=42)
x_umap2d = umap2d.fit_transform(x_pca)

# ================================
# UMAP in 3D
# ================================
print("Running 3D UMAP...")
umap3d = UMAP(n_neighbors=30, min_dist=0.5, n_components=3, random_state=42)
x_umap3d = umap3d.fit_transform(x_pca)

# ================================
# Interactive Plotly 2D
# ================================
df_2d = pd.DataFrame({
    "UMAP1": x_umap2d[:,0],
    "UMAP2": x_umap2d[:,1],
    "Label": [class_names[i] for i in y]
})

fig2d = px.scatter(
    df_2d, x="UMAP1", y="UMAP2",
    color="Label", title="2D Feature Space (UMAP)",
    opacity=0.6, width=900, height=700
)
fig2d.show()

# ================================
# Interactive Plotly 3D
# ================================
df_3d = pd.DataFrame({
    "UMAP1": x_umap3d[:,0],
    "UMAP2": x_umap3d[:,1],
    "UMAP3": x_umap3d[:,2],
    "Label": [class_names[i] for i in y]
})

fig3d = px.scatter_3d(
    df_3d, x="UMAP1", y="UMAP2", z="UMAP3",
    color="Label", title="3D Feature Space (UMAP)",
    opacity=0.6, width=900, height=700
)

# Show interactive 3D plot
fig3d.show()

# ================================
# Save 3D plot as HTML
# ================================
fig3d.write_html("feature_space_umap3d.html")
print("✅ 3D UMAP plot saved as feature_space_umap3d.html")
```
